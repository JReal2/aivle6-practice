```python
import numpy as np

# 각 그룹마다 랜덤값 부여
# Group A는 100~200 실수값
# Group B는 0~50
# Group C는 x축은 100~200, y축은 0~50
group_A = np.random.uniform(100, 200, (15, 2))
group_B = np.random.uniform(0, 50, (15, 2))
group_C = np.column_stack((np.random.uniform(100, 200, 15), np.random.uniform(0, 50, 15)))

#각 그룹의 점들과 우리가 갖고 있는 현재 point와의 거리를 구하는 함수
def get_distance(p, a):
  return np.sqrt(np.sum((p - a) ** 2))

k = 5

p = np.random.uniform(0, 200, 2)
print(p)

distances = []

for index, group in enumerate([group_A, group_B, group_C]):
    for point in group:
        distance = get_distance(p, point)
        distances.append((distance, index))

distances.sort()

# 그룹별로 가까운 점 개수 세기
stat = [0, 0, 0]

for _, group_index in distances[:k]:
    stat[group_index] += 1

# 최대로 큰 값의 위치를 구해 해당 그룹 출력
index = np.argmax(stat)
if index == 0:
    print("Group A")
elif index == 1:
    print("Group B")
else:
    print("Group C")
```
