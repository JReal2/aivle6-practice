# 1.환경 준비

- 기본 라이브러리와 대상 데이터를 가져와 이후 과정을 준비합니다.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings

warnings.filterwarnings(action='ignore')
```

```python
# https://raw.githubusercontent.com/jangrae/csv/master/Attrition_simple2.csv
path = 'https://raw.githubusercontent.com/jangrae/csv/master/Attrition_simple2.csv'
data = pd.read_csv(path)
```

# 2.데이터 이해

- 분석할 데이터를 충분히 이해할 수 있도록 다양한 탐색 과정을 수행합니다.

# 3.데이터 준비

- 전처리 과정을 통해 머신러닝 알고리즘에 사용할 수 있는 형태의 데이터를 준비합니다.

**1) 변수 제거**

```python
# EmployeeNumber
drop_cols = 'EmployeeNumber'

data.drop(columns=drop_cols, inplace=True)
```

**2) x, y 분리**

```python
target = 'Attrition'

x = data.drop(columns=target)
y = data[target]
```

**2) 가변수화**

```python
dumm_cols = ['Gender', 'MaritalStatus', 'OverTime']

x = pd.get_dummies(x, columns=dumm_cols, drop_first=True, dtype=int)
```

**3) 학습용, 평가용 데이터 분리**

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=1)
```

**4) 정규화**

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
x_train_s = scaler.fit_transform(x_train)
x_test_s = scaler.transform(x_test)
```

# 4.모델링

- 하이퍼파라미터 최적화 과정을 통해 최선의 성능을 갖는 모델을 만들고 성능을 검증합니다.
- 각 모델을 joblib.dump() 함수를 사용해 저장합니다.

- 이후 사용할 함수를 모두 불러옵니다.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import *
```

**1) KNN**

- 최적으로 파라미터로 학습된 KNN 모델을 만들고, 최고의 파라미터와 성능을 확인합니다.
- 모델 이름은 model_knn로 지정합니다.

```python
param = {'n_neighbors' : range(1, 51)}
model = RandomizedSearchCV(KNeighborsClassifier(),
                          param,
                          cv=5)
model.fit(x_train, y_train)

print(model.best_params_)
print(model.best_score_)

result={}
result['KNN'] = model.best_score_
```

```python
joblib.dump(model, 'model_knn.pkl')
```

**2) Decision Tree**

- 최적으로 파라미터로 학습된 Decision Tree 모델을 만들고, 최고의 파라미터와 성능을 확인합니다.
- 모델 이름은 model_dst로 지정합니다.

```python
param = {'max_depth' : range(1, 51)}
model = RandomizedSearchCV(DecisionTreeClassifier(),
                          param,
                          cv=5)

model.fit(x_train, y_train)

print(model.best_params_)
print(model.best_score_)

result['Decision Tree'] = model.best_score_
```

```python
joblib.dump(model, 'model_dst.pkl')
```

**3) Random Forest**

- 최적으로 파라미터로 학습된 Random Forest 모델을 만들고, 최고의 파라미터와 성능을 확인합니다.
- 모델 이름은 model_rdf로 지정합니다.

```python
param = {'max_depth' : range(1, 51)}
model = RandomizedSearchCV(RandomForestClassifier(),
                          param,
                          cv=5)

model.fit(x_train, y_train)

print(model.best_params_)
print(model.best_score_)

result['Random Forest'] = model.best_score_
```

```python
joblib.dump(model, 'model_rdf.pkl')
```

**4) LightGBM**

- 최적으로 파라미터로 학습된 LightGBM 모델을 만들고, 최고의 파라미터와 성능을 확인합니다.
- 모델 이름은 model_lgb로 지정합니다.

```python
param = {'max_depth' : range(1, 51)}
model = RandomizedSearchCV(LGBMClassifier(),
                          param,
                          cv=5)

model.fit(x_train, y_train)

print(model.best_params_)
print(model.best_score_)

result['LightGBM'] = model.best_score_
```

```python
joblib.dump(model, 'model_lgb.pkl')
```

```python
plt.barh(list(result), result.values())
```
