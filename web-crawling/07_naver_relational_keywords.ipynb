### 네이버 함께 많이 찾는 주제어 수집
- 정적(static) 웹페이지 데이터 수집 
- BeautifulSoup을 이용하여 HTML 문자열 데이터 parsing

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

#### 1. 웹페이지 분석 : URL

```python
url = 'https://search.naver.com/search.naver?query=%EC%82%BC%EC%84%B1%EC%A0%84%EC%9E%90'
```

#### 2. request(URL) > response : str(html)

```python
response = requests.get(url)
response
```

#### 3. str(html) > bs object

```python
dom = BeautifulSoup(response.text,'html.parser')
type(dom)
```

#### 4. bs object > .select(css-selector), .select_one(css-selector) > str(text)

```python
#fdr-293948c105b74eb1817d8ced4815c775 > div > div > div.bAUlEC095ho5bG7U0km1.fds-grid-layout-keyword.fds-refine-query-grid > div:nth-child(1) > div:nth-child(1) > a > span
```

```python
# select > list(tag1, tag2, ...), select_one > tag
```

```python
elements = dom.select('.fds-refine-query-grid span')
len(elements), elements
```

#### 5. str(text) > DataFrame

```python
elements[0].text
```

```python
[element.text for element in elements]
```
