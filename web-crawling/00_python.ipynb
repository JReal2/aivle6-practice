# Class
- 변수와 함수를 묶어서 코드 작성 및 실행
- 객체지향 구현 : 실제세계를 모델링하여 프로그램을 개발하는 개발 방법론 - 협업이 용이함
- 함수 사용 : 함수 선언(코드 작성) -> 함수 호출(코드 실행)
- 클래스 사용
   - 클래스 선언(코드 작성 : 설계도 작성) -> 객체 생성(메모리 사용 : 제품 생산) -> 메서드 실행(코드 실행 : 제품 사용)
   - 메서드 : 클래스 안에 선언되는 함수
- Class 식별자 : UpperCamelCase, PascalCase(0), snake_case()

```python
# 1. 클래스 선언 : 코드 작성
# 예금 계좌 : Account 클래스 - balance, deposit(), withdraw()

class Account:
    balance = 0
    
    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if (self.balance >= amount):
            self.balance -= amount
```

```python
# 2. 객체 생성 : 메모리 사용
acc1 = Account()
acc2 = Account()
```

```python
# dir() : 객체 안에 들어있는 변수, 함수 목록 출력
dir(acc1)[-3:], dir(acc2)[-3:]
```

```python
acc1.balance, acc2.balance
```

```python
# 3. 메서드 실행 : 코드 실행
```

```python
acc1.deposit(1000)
acc1.withdraw(600)
acc2.deposit(3000)
```

```python
acc1.balance, acc2.balance
```

### special methods
- 특별한 기능을 하는 메서드
- 앞 뒤로 __가 붙은 메서드
- 대표적으로 생성자 메서드

```python
# 생성자 메서드 : __init__(self)
# 객체 생성 시 실행되는 메서드
```

```python
# 클래스선언
class Account:
    # 생성자 메서드 : 메서드에서 사용되는 변수와 초기값 설정하는 용도로 사용
    def __init__(self, balance):
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount):
        self.balance -= amount
```

```python
acc = Account(1000)
```

```python
acc.deposit(1000)
```

```python
acc.balance
```

```python
class Number:
    def __init__(self, data):
        self.data = data
    def __add__(self, obj):
        self.data -= obj.data
        return self.data
```

```python
num1 = Number(3)
num2 = Number(2)
```

```python
num2.data
```

```python
num1 + num2 # num1__add__(num2)
```
