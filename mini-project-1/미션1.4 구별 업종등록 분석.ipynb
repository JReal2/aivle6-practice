## [실습4] 시각화 분석

- 다양한 열를 기준으로 그래프를 그려보고 인사이트를 도출해보세요.

```python
# 시각화 import
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 자치구별 업종 분포
plt.plot(seoul_business['자치구'], seoul_business['한식 일반 음식점업'], label = '한식 일반 음식점업')
plt.plot(seoul_business['자치구'], seoul_business['부동산 중개 및 대리업'], label = '부동산 중개 및 대리업')
plt.plot(seoul_business['자치구'], seoul_business['커피전문점'], label = '커피전문점')
plt.plot(seoul_business['자치구'], seoul_business['기타주점업'], label = '기타주점업')
plt.plot(seoul_business['자치구'], seoul_business['일반 교과 학원'], label = '일반 교과 학원')

plt.xticks(rotation=90)
plt.xlabel('자치구')
plt.legend(loc = 'upper left')

plt.show()
```

```python
plt.figure(figsize = (10, 5))
plt.xlabel('자치구')
plt.ylabel('업종별 종사자 수')
plt.bar(seoul_business['자치구'], seoul_business['한식 일반 음식점업'], label = '한식 일반 음식점업')
plt.bar(seoul_business['자치구'], seoul_business['부동산 중개 및 대리업'], bottom = seoul_business['한식 일반 음식점업'], label = '부동산 중개 및 대리업')
plt.bar(seoul_business['자치구'], seoul_business['커피전문점'], bottom = seoul_business['한식 일반 음식점업'] + seoul_business['부동산 중개 및 대리업'], label = '커피전문점')
plt.bar(seoul_business['자치구'], seoul_business['기타주점업'], bottom = seoul_business['한식 일반 음식점업'] + seoul_business['부동산 중개 및 대리업'] + seoul_business['커피전문점'], label = '기타주점업')
plt.bar(seoul_business['자치구'], seoul_business['일반 교과 학원'], bottom = seoul_business['한식 일반 음식점업'] + seoul_business['부동산 중개 및 대리업'] + seoul_business['커피전문점'] + seoul_business['기타주점업'], label = '일반 교과 학원')
plt.legend(loc = 'upper left')
plt.xticks(rotation=90)
plt.show()
```
