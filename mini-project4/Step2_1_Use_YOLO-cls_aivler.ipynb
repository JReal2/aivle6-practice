#### 1) 본인 얼굴 이미지 데이터셋 불러오기

```python
import zipfile
import os

## Colab에 생성할 다른 얼굴 폴더 경로
extract_folder = '/content/my_face'

## 위의 경로에 폴더가 없을 때 생성
if not os.path.exists(extract_folder) :
    os.makedirs(extract_folder)

## 위의 경로에 압축을 해제
with zipfile.ZipFile(path + '/Datasets/Keras/my_face.zip', 'r') as zip_ref :
    file_list = zip_ref.namelist()

    for f in file_list :
        if not f.endswith('/') and f.lower().endswith('.jpg') :
            file_name = os.path.basename(f)

            if not file_name.startswith('._') :
                d_path = os.path.join(extract_folder, file_name)

                with zip_ref.open(f) as source, open(d_path, 'wb') as target :
                    target.write(source.read())
```

#### 2) 다른 얼굴 이미지 데이터셋 불러오기

```python
## Colab에 생성할 다른 얼굴 폴더 경로
extract_folder = '/content/other_face'

## 위의 경로에 폴더가 없을 때 생성
if not os.path.exists(extract_folder) :
    os.makedirs(extract_folder)

## 위의 경로에 압축을 해제
with zipfile.ZipFile(path + '/Datasets/Keras/lfw-deepfunneled.zip', 'r') as zip_ref :
    file_list = zip_ref.namelist()

    for f in file_list :
        if not f.endswith('/') and f.lower().endswith('.jpg') :
            file_name = os.path.basename(f)

            if not file_name.startswith('._') :
                d_path = os.path.join(extract_folder, file_name)

                with zip_ref.open(f) as source, open(d_path, 'wb') as target :
                    target.write(source.read())
```

### (2) 데이터셋 전처리

* **세부 요구사항**
    - 데이터셋을 전처리 합니다.
        - YOLO-cls 모델이 요구하는 폴더 구조를 만듭니다.
            1. Datasets라는 폴더를 생성합니다.
            2. Training set, Validation set, Test set(선택 사항) 각 데이터셋이 들어갈 폴더를 생성합니다.
            3. 각 데이터셋 폴더에 분류할 클래스의 이름을 가진 폴더를 생성합니다.
        - 폴더 구조에 맞게 데이터를 분배합니다.
    - 예시 코드에서 사용한 라이브러리
        - os, glob, random, shutil, numpy

#### 1) 모델이 요하는 구조의 폴더 생성

```python
import random

base_dir = '/content'

train_dir = os.path.join(base_dir, 'train')
val_dir = os.path.join(base_dir, 'val')
test_dir = os.path.join(base_dir, 'test')

for folder in [train_dir, val_dir, test_dir]:
    os.makedirs(os.path.join(folder, 'my'), exist_ok=True)
    os.makedirs(os.path.join(folder, 'other'), exist_ok=True)
```

#### 2) 각 폴더에 이미지 데이터 옮기기

```python
import shutil

my_face_dir = base_dir+'/my_face'
other_face_dir = base_dir+'/other_face'

my_images = os.listdir(my_face_dir)
other_images = os.listdir(other_face_dir)

random.seed(2024)
random.shuffle(my_images)
random.shuffle(other_images)

num_my = len(my_images)
num_other = len(other_images)

for i, img in enumerate(my_images):
    src_path = os.path.join(my_face_dir, img)
    if i < num_my * 7 / 10:
        dst_path = os.path.join(train_dir, 'my', img)
    elif i < num_my * 9 / 10:
        dst_path = os.path.join(val_dir, 'my', img)
    else:
        dst_path = os.path.join(test_dir, 'my', img)
    shutil.copy(src_path, dst_path)

for i, img in enumerate(other_images):
    src_path = os.path.join(other_face_dir, img)
    if i < num_other * 7 / 10:
        dst_path = os.path.join(train_dir, 'other', img)
    elif i < num_other * 9 / 10:
        dst_path = os.path.join(val_dir, 'other', img)
    else:
        dst_path = os.path.join(test_dir, 'other', img)
    shutil.copy(src_path, dst_path)
```
