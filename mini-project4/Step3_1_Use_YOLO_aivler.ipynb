### (2) 데이터셋 전처리

* **세부 요구사항**
    - Label 파일, 즉 txt 파일의 클래스를 수정하셔야 합니다.
        - 여러 클래스로 구분되어 있는 것을 하나로 통일해주세요.
        - 이것은 **다른 사람의 얼굴**에 대해 전처리하는 작업입니다.
    - UltraLytics YOLO 모델에서 요구하는 데이터셋 폴더의 구조가 있습니다.
    - [UltraLytics YOLO 모델의 데이터셋 구조 링크](https://docs.ultralytics.com/datasets/detect/)
    - 예시 코드에서 사용한 라이브러리
        - os, glob, random, shutil, numpy

#### 1) Label 파일의 클래스 값 변경

```python
import glob

txt_files = glob.glob('images/**/labels/*.txt')

for file_path in txt_files:
  with open(file_path, 'r') as file:
    lines = file.readlines()

    updated_lines = []
    for line in lines:
        parts = line.strip().split()
        parts[0] = '0'
        updated_lines.append(' '.join(parts))

    # 변경된 내용을 파일에 다시 씁니다
    with open(file_path, 'w') as file:
        file.write('\n'.join(updated_lines))
```

```python
txt_files = glob.glob('my_face/*.txt')

for file_path in txt_files:
  with open(file_path, 'r') as file:
    lines = file.readlines()

    updated_lines = []
    for line in lines:
        parts = line.strip().split()
        parts[0] = '1'
        updated_lines.append(' '.join(parts))

    with open(file_path, 'w') as file:
      file.write('\n'.join(updated_lines))
```

#### 2) 모델이 요하는 구조의 폴더 생성

```python
base_dir = '/content/my_face'
data_dir = '/content/images'

train_dir = os.path.join(data_dir, 'train')
valid_dir = os.path.join(data_dir, 'valid')
test_dir = os.path.join(data_dir, 'test')
```

#### 3) 각 폴더에 데이터 옮기기

```python
my_images = sorted(glob.glob(os.path.join(base_dir, '*.jpg')))
my_labels = sorted(glob.glob(os.path.join(base_dir, '*.txt')))

random.seed(2024)
random.shuffle(my_images)
random.shuffle(my_labels)

num_my = len(my_images)

for i, img_path in enumerate(my_images):
  label_path = img_path.replace('.jpg', '.txt')

  if i < num_my * 7 / 10:
      dst_image_dir = os.path.join(train_dir, 'images')
      dst_label_dir = os.path.join(train_dir, 'labels')
  elif i < num_my * 9 / 10:
      dst_image_dir = os.path.join(valid_dir, 'images')
      dst_label_dir = os.path.join(valid_dir, 'labels')
  else:
      dst_image_dir = os.path.join(test_dir, 'images')
      dst_label_dir = os.path.join(test_dir, 'labels')

  shutil.move(img_path, os.path.join(dst_image_dir, os.path.basename(img_path)))
  shutil.move(label_path, os.path.join(dst_label_dir, os.path.basename(label_path)))
```

```python
for folder in ['train', 'valid', 'test']:
    image_folder = os.path.join(data_dir, folder, 'images')
    image_files = glob.glob(os.path.join(image_folder, '*.jpg'))

    dest_folder = os.path.join(data_dir, folder)
    os.makedirs(dest_folder, exist_ok=True)

    for label_file in image_files:
        shutil.move(label_file, dest_folder)

    shutil.rmtree(image_folder)
```

```python
label_dest_dir = '/content/labels'

for folder in ['train', 'valid', 'test']:
    label_folder = os.path.join(data_dir, folder, 'labels')
    label_files = glob.glob(os.path.join(label_folder, '*.txt'))

    dest_folder = os.path.join(label_dest_dir, folder)
    os.makedirs(dest_folder, exist_ok=True)

    for label_file in label_files:
        shutil.move(label_file, dest_folder)

    shutil.rmtree(label_folder)
```
